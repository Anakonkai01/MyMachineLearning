Quy ước chuẩn trong sách D2L (và Toán học cao cấp):

    Order (Bậc): Dùng để chỉ số lượng trục (số lượng chỉ số index cần dùng để truy cập phần tử).

        Vector có Order = 1 (1 trục).

        Matrix có Order = 2 (2 trục: hàng, cột).

        Tensor 3D có Order = 3.

    Dimensionality (Số chiều/Độ dài): Dùng để chỉ số lượng phần tử (components) bên trong vector đó.

```python
# tranpose 
A = torch.randn(3,3)
print(A)

A == A.T
print(A * A**-1 )
print(A**-1 * A)
```

# testing with reducing

```python
A = torch.arange(12, dtype= float).reshape(3,-1)
print(A) 
print(A.shape)
print(A.sum(axis=0))
print(A.sum(axis=0).shape)


# testing with reducing axis = 1 

A = torch.arange(3*4*5*6*7, dtype=float).reshape(3,4,5,6,-1)
print(A)
print(A.shape)
print(A.sum(axis=0))
print(A.sum(axis=0).shape)
```

```python
# get the average 
A.mean(), A.sum() / A.numel()
```

```python
# test between numel and len 
print(len(A)) # this only count for first axis = 0
print(A.numel())
```

```python
# non-reduction sum 
A = torch.arange(9, dtype=float).reshape(3,-1)
sum_A = A.sum(axis=0, keepdim=True)
print(sum_A)
print(sum_A.shape)


# this use for broadcasting
# ex: 
A / sum_A
```

# NORM 

‖αx‖ = |α| · ‖x‖


‖x + y‖ ≤ ‖x‖ + ‖y‖


A norm is a function

that maps a vector to a scalar and satisfies the following three properties:

    Given any vector 

, if we scale (all elements of) the vector by a scalar

, its norm scales accordingly:
(2.3.10)

For any vectors
and

: norms satisfy the triangle inequality:
(2.3.11)

The norm of a vector is nonnegative and it only vanishes if the vector is zero:
(2.3.12)


```python
# norm, default of pytorch is l2 norm 
u = torch.tensor([3.0, -4.0])

l2_norm_u = torch.norm(u)

print(l2_norm_u)
```

```python
# custom l1 norm 

l1_norm_u = torch.abs(torch.sum(u))

print(l1_norm_u)
```

```python
# if we use norm, it mean apply frobenius norm 
torch.norm(torch.ones(3,4))

```
