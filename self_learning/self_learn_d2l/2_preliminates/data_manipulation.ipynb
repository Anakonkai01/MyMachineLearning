# Operations

## Elementwise Operations 

```python
# elementwise exp 

# unary - 1 input -> 1 output 
print(torch.exp(x))
print(x ** 2)
print(torch.sin(x))
print(torch.abs(x))


# binary - 2 input -> 1 output
# make that x and y has have the same shape
y = 2 * x
print(x * y)
print(x*2) # this still binary but it automatic compute elementwise =)) 
print(x+10)
```

```python
# concatenate 
# ex perform concate with axis = 0

X = torch.arange(10).reshape(2,-1)
Y = torch.arange(10,20).reshape(2,-1)
print(X)
print(Y)


# perform concat 
Z = torch.cat((X, Y), dim=1)
print(Z)

# condition to cat 
# cat dim = 0, dim = 1 must equal, and vice versa 

```

# broadcasting 

1. Quy tắc hoạt động (The Rules)

Để hai Tensor A và B có thể "Broadcast" được với nhau, chúng phải tuân thủ quy tắc nghiêm ngặt:

Quy tắc: So sánh kích thước (shape) của chúng từ Phải sang Trái (Right-to-Left). Tại mỗi chiều, hai kích thước phải thỏa mãn 1 trong 2 điều kiện:

    Chúng bằng nhau.

    Một trong hai kích thước bằng 1.

Nếu kích thước bị thiếu (ví dụ A là 3D, B là 1D), ta coi như kích thước bị thiếu là 1.

```python
# the broadcasting auto stretching the another tensor to match shape 
A = torch.arange(6).reshape(3,-1)
print(A)
b = torch.arange(2).reshape(1,-1)
print(b)

res = A + b 
print(res)
```
