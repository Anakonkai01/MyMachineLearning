# yield
- khác với return, nó sẽ ko ngắt toàn bộ function đó, mà nó sẽ pause lại, và trả về kết quả, khi cần dùng tiếp nó sẽ tiếp tục

```python
# generator là gì đối tượng iterator trong python nó lưu trữ tất cả giá trị trong bộ nhớ mà tạo ra giá trị tiếp theo chỉ khi được yêu cầu 

def count():
    yield 1
    yield 2
    yield 3
    yield 4
    
gen = count()

print(next(gen))
print(next(gen))
```

```python
# using for mini batch 
def get_batches(data, batch_size):
    for i in range(0, len(data), batch_size):
        batch = data[i: i + batch_size]
        yield batch
        
        
def train_on(minibatch):
    return "hello"

import torch
mydata = torch.randn(1000, 2)
for minibatch in get_batches(mydata, 32):
    # train on mini batch 
    train_on(minibatch) # only 32 batch in ram 
```

# iterable, iterator

```python
import random 

def may_goi_so(n):
    for i in range(1, n + 1):
        yield i

may = may_goi_so(1000) # this is iterable 

# the iterator 
a = iter(may)
b = iter(may )

print(next(a))
print(next(b))

```
