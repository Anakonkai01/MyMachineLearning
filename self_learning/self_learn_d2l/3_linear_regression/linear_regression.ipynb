```python
%matplotlib inline
import math
import time
import numpy as np 
import torch
```

```python
# vectorization over for loop to maximize speed 
# ex: vector a + vector b 
n = 10000
a = torch.ones(n)
b = torch.ones(n)
```

```python
c = torch.zeros(n)
t1 = time.time()
for i in range(n):
    c[i] = a[i] + b[i]
t2 = time.time()
```

```python
# vectorization
t3 = time.time()
d = a + b
t4 = time.time()

print("for loop time", t2-t1)
print("vectorization", t4-t3)
```
