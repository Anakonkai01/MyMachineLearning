```python
import torch 
```

```python
# gia lap i = v / r 
I = torch.tensor([1.0, 2.0, 3.0, 4.0]) # Dòng điện
V_true = torch.tensor([5.0, 10.0, 15.0, 20.0]) # Hiệu điện thế đo được

R = torch.randn(1, requires_grad=True)

lr = 0.01 

for epoch in range(100):
    # forward 
    V_pred = I * R 
    # compute loss 
    loss = torch.mean((V_true - V_pred) ** 2)
    # backward 
    loss.backward()
    # update 
    with torch.no_grad():
        R -= lr * R.grad
    # reset grad 
    R.grad.zero_()

print("R", R.item())
```
